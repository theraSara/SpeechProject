# Official-split analysis of classical baselines

This notebook summarizes the official-split results for the two classical baselines used in the project:

- **MFCC + DTW**
- **MFCC + HMM-GMM**

The evaluation follows the shared 5-shot and 10-shot episodic splits and reports the agreed operating-point metrics:

- **ACC+@5% FAR**
- **AUROC**
- **FRR+@5% FAR**

The focus is on concise figures that can be reused in the individual portfolio report and the group report.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
outdir = Path('.')
summary_df = pd.read_csv('results/baseline_results_table.csv')
per_trial_df = pd.read_csv('official_per_trial_metrics.csv')
summary_df[['method','acc_plus_5','auroc','frr_plus_5','far_at_5','mean_inference_time_ms']]
```

## Main observations

1. **HMM-GMM consistently outperforms DTW** under the official fixed-FAR evaluation.
2. The best overall baseline is **MFCC+HMM-GMM (10-shot)** with:
   - ACC+@5% = **22.09%**
   - AUROC = **0.726**
   - FRR+@5% = **69.30%**
3. DTW shows **very small gains from 5-shot to 10-shot**, while latency and storage roughly double.
4. HMM-GMM benefits substantially from 10-shot enrollment and is also **faster and smaller** than DTW in this implementation.
